````python
# ============================================================
# 第 1 格：安裝套件、讀取金鑰、設定模型（請勿修改）
# ============================================================
!pip install -q groq gradio

import os
import re
import time
import sqlite3
import pandas as pd
from IPython.display import display
from groq import Groq
import gradio as gr

# 判斷目前是在 Colab 還是本地端 Jupyter
try:
    from google.colab import userdata
    IN_COLAB = True
except ImportError:
    IN_COLAB = False

# 🔴 金鑰名稱一律為 GROQ_API_KEY；絕對不可以把金鑰寫在程式裡，也不可以 print 出來
if IN_COLAB:
    try:
        GROQ_API_KEY = userdata.get('GROQ_API_KEY')
    except Exception:
        print("❌ 讀不到金鑰：請到左側 🔑 新增名稱為 GROQ_API_KEY 的密鑰，並打開「筆記本存取權」")
        raise
else:
    GROQ_API_KEY = os.environ.get('GROQ_API_KEY')
    if not GROQ_API_KEY:
        raise RuntimeError("找不到環境變數 GROQ_API_KEY")

client = Groq(api_key=GROQ_API_KEY)

# 期中考指定模型；推理強度設為 low，可大幅減少 token 用量
# （唯一允許修改處：若 20b 額度用完，可改為 "openai/gpt-oss-120b"，並在規格書最後註明）
MODEL = "openai/gpt-oss-20b"
REASONING_EFFORT = "low"
print("✅ 環境：", "Colab" if IN_COLAB else "本地端", "｜模型：", MODEL)
````

````python
# ============================================================
# 第 2 格：建立資料庫與初始資料
# 提醒：每次執行都要 DROP TABLE IF EXISTS 再重建
# ============================================================
# （把 AI 產生、你看過並理解的程式貼在這裡）

# ============================================================
# 第 2 格：建立 SQLite 資料庫與初始資料
# ============================================================

# 設定資料庫檔案名稱
DB_NAME = "activity_registration.db"

conn = None

try:
    # 1. 連接 SQLite 資料庫
    conn = sqlite3.connect(DB_NAME)
    cursor = conn.cursor()

    cursor.execute("PRAGMA foreign_keys = ON")

    # 2. 刪除舊資料表，確保每次執行都重新初始化
    # 先刪除子表，再刪除主表
    cursor.execute("DROP TABLE IF EXISTS registrations")
    cursor.execute("DROP TABLE IF EXISTS events")

    # 3. 建立活動資料表
    cursor.execute("""
        CREATE TABLE events (
            id INTEGER PRIMARY KEY AUTOINCREMENT,
            name TEXT UNIQUE NOT NULL,
            event_date TEXT NOT NULL,
            capacity INTEGER NOT NULL CHECK (capacity >= 0)
        )
    """)

    # 4. 建立報名紀錄資料表
    cursor.execute("""
        CREATE TABLE registrations (
            id INTEGER PRIMARY KEY AUTOINCREMENT,
            event_id INTEGER NOT NULL,
            student_name TEXT NOT NULL,
            people INTEGER NOT NULL
                CHECK (people BETWEEN 1 AND 3),
            status TEXT NOT NULL DEFAULT '已報名'
                CHECK (status IN ('已報名', '已取消')),
            created_at TIMESTAMP DEFAULT CURRENT_TIMESTAMP,

            FOREIGN KEY (event_id) REFERENCES events(id),
            UNIQUE (event_id, student_name)
        )
    """)

    # 5. 三筆初始活動
    initial_events = [
        ("迎新晚會", "2026-11-20", 6),
        ("程式工作坊", "2026-11-25", 4),
        ("籃球友誼賽", "2026-11-28", 3)
    ]

    # 6. 新增初始活動
    cursor.executemany(
        """
        INSERT INTO events (name, event_date, capacity)
        VALUES (?, ?, ?)
        """,
        initial_events
    )

    # 7. 儲存所有變更
    conn.commit()

    print("✅ 資料庫建立成功！")
    print("✅ events 活動資料表建立完成！")
    print("✅ registrations 報名紀錄表建立完成！")
    print(f"✅ 已新增 {len(initial_events)} 筆初始活動！")

except sqlite3.Error as e:
    # 發生錯誤時取消尚未提交的操作
    if conn is not None:
        conn.rollback()

    print(f"❌ 資料庫建立失敗：{e}")
    raise

finally:
    # 8. 確保資料庫連線關閉
    if conn is not None:
        conn.close()
````

````python
# ============================================================
# 第 3 格：顯示資料表
# 提醒：定義 show_tables() 並呼叫一次
# ============================================================
# （把 AI 產生、你看過並理解的程式貼在這裡）

# ============================================================
# 第 3 格：顯示資料表
# 功能：顯示活動資料與報名紀錄
# ============================================================

def show_tables():

    # 1. 連接 SQLite
    conn = sqlite3.connect(DB_NAME)

    try:
        # 2. 查詢活動資料表
        events_df = pd.read_sql_query(
            """
            SELECT id, name, event_date, capacity
            FROM events
            ORDER BY id
            """,
            conn
        )

        # 3. 查詢報名紀錄資料表
        registrations_df = pd.read_sql_query(
            """
            SELECT id, event_id, student_name,
                   people, status, created_at
            FROM registrations
            ORDER BY id
            """,
            conn
        )

        # 4. 顯示活動資料
        print("===== 活動資料表 events =====")
        display(events_df)

        # 5. 顯示報名紀錄
        print("\n===== 報名紀錄表 registrations =====")

        display(registrations_df)

        if registrations_df.empty:
            print("目前尚無報名紀錄。")

    finally:
        # 6. 確保資料庫連線關閉
        conn.close()

show_tables()
````

````python
# ============================================================
# 第 4 格：AI 解讀函式
# 提醒：回傳 (動作, 活動, 人數, 原始回覆)；不可直接相信 AI 的回覆
# ============================================================
# （把 AI 產生、你看過並理解的程式貼在這裡）

# 1. 設定允許的動作與活動
ACTIONS = ["報名", "取消", "查詢", "未知"]

EVENT_NAMES = [
    "迎新晚會",
    "程式工作坊",
    "籃球友誼賽"
]

ALLOWED_EVENTS = EVENT_NAMES + ["全部活動", "未知"]

# 2. 建立快取，避免相同輸入重複呼叫 AI
ai_cache = {}


# 3. 處理學生輸入中明確提到的人數
def extract_people(message, default=1):

    # 例如：共4人、總共3人、報名2人
    match = re.search(
        r"(?:共|總共|報名)\s*(\d+)\s*人",
        message
    )
    if match:
        return int(match.group(1))

    # 例如：我和兩個朋友、我想和1個朋友
    match = re.search(
        r"(?:和|帶)\s*([一二兩三四五六七八九\d]+)\s*個?朋友",
        message
    )

    if match:
        value = match.group(1)

        chinese_numbers = {
            "一": 1, "二": 2, "兩": 2,
            "三": 3, "四": 4, "五": 5,
            "六": 6, "七": 7, "八": 8,
            "九": 9
        }

        if value.isdigit():
            return int(value) + 1

        if value in chinese_numbers:
            return chinese_numbers[value] + 1

    return default


# 4. 建立 AI 解讀函式
def ai_parse(message):

    # 檢查空白與非文字輸入
    if not isinstance(message, str) or not message.strip():
        return ("未知", "未知", 1, "輸入為空白")

    message = message.strip()

    # 檢查快取
    if message in ai_cache:
        return ai_cache[message]

    # 5. 設定 AI 辨識規則
    prompt = """
你是系學會活動報名系統的 AI 文字辨識助手。

目前有三個活動：
1. 迎新晚會
2. 程式工作坊
3. 籃球友誼賽

請將學生輸入的話轉換成以下格式：

動作,活動,人數

【動作規則】
只能使用：報名、取消、查詢、未知。

「我要參加」或「我想報名」代表報名。
「取消報名」代表取消。
「剩餘名額」或「還有幾個名額」代表查詢。
「各活動報名狀況」或「所有活動統計」
代表查詢全部活動。

【活動規則】
只能使用：
迎新晚會、程式工作坊、籃球友誼賽、
全部活動、未知。

若學生提到不存在的活動，活動回傳未知。
若學生同時要求多個活動，活動回傳未知。
若完全無法判斷動作，動作回傳未知。

【人數規則】
未指定人數時，預設為1。
「我和一個朋友」代表2人。
「我和兩個朋友」代表3人。
「我和三個朋友」代表4人。
「共4人」代表4人。

即使人數超過3，也要回傳原本的人數。
不可自行將超標人數改成3。
人數是否符合規定由後續 Python 程式判斷。

【輸出要求】
只能輸出一行，使用半形逗號分隔。
不能加上解釋、標題或其他文字。

範例：
我想報名迎新晚會。
報名,迎新晚會,1

我和兩個朋友要報名程式工作坊。
報名,程式工作坊,3

我要取消籃球友誼賽的報名。
取消,籃球友誼賽,1

查詢各活動報名狀況。
查詢,全部活動,1

我要報名程式工作坊，共4人。
報名,程式工作坊,4

我想報名不存在的演唱會。
報名,未知,1

你好，今天天氣很好。
未知,未知,1
"""

    raw_reply = ""

    # 6. 呼叫 Groq API，最多嘗試3次
    for attempt in range(3):

        try:
            response = client.chat.completions.create(
                model=MODEL,
                messages=[
                    {"role": "system", "content": prompt},
                    {"role": "user", "content": message}
                ],
                temperature=0,
                reasoning_effort=REASONING_EFFORT
            )

            # 取得 AI 原始回覆
            raw_reply = (
                response.choices[0].message.content or ""
            ).strip()

            break

        except Exception as e:

            error_code = getattr(e, "status_code", None)

            # 遇到429：等待30秒後重試
            if error_code == 429 and attempt < 2:
                print(
                    f"⚠️ API 額度限制，第{attempt + 1}次失敗，"
                    "30秒後重試..."
                )
                time.sleep(30)

            else:
                print(f"❌ AI 呼叫失敗：{type(e).__name__}")
                return (
                    "未知", "未知", 1,
                    f"API 呼叫失敗：{type(e).__name__}"
                )

        finally:
            # 每一次 API 呼叫後都暫停2秒
            time.sleep(2)

    # 7. 解析 AI 回覆
    # 支援半形及全形逗號
    parts = [
        p.strip()
        for p in re.split(r"[,，]", raw_reply)
    ]

    action = "未知"
    event = "未知"
    people = 1

    # 從允許清單驗證動作
    if len(parts) >= 1:
        if parts[0] in ACTIONS:
            action = parts[0]

    # 從允許清單驗證活動
    if len(parts) >= 2:
        if parts[1] in ALLOWED_EVENTS:
            event = parts[1]

    # 用正規表示式擷取人數
    if len(parts) >= 3:
        match = re.search(r"\d+", parts[2])

        if match:
            people = int(match.group())

    # 8. 補充檢查學生原始輸入的人數
    # 若有明確的攜伴或總人數，以學生原話為準
    original_people = extract_people(message)

    if original_people != 1:
        people = original_people

    # 9. 防止一次指定多種活動
    mentioned_events = [
        name for name in EVENT_NAMES
        if name in message
    ]

    if len(mentioned_events) > 1:
        event = "未知"

    # 10. 儲存結果至快取
    result = (action, event, people, raw_reply)

    ai_cache[message] = result

    return result
````

````python
# ============================================================
# 第 6 格：功能函式
# 提醒：fetchone() 後先檢查 None；用 try/finally 關閉連線；統計用 JOIN＋GROUP BY
# ============================================================
# （把 AI 產生、你看過並理解的程式貼在這裡）

# ============================================================
# 第 6 格：系學會活動報名系統－核心功能
# 包含：報名、取消、查詢、A1、A2、A3
# ============================================================

# 1. 建立活動統計函式（A3）
def event_statistics():

    conn = sqlite3.connect(DB_NAME)

    try:
        cursor = conn.cursor()

        # LEFT JOIN：確保沒有人報名的活動也會顯示
        # GROUP BY：依活動統計實際報名人數
        cursor.execute("""
            SELECT
                e.name,
                e.event_date,
                e.capacity,
                COALESCE(
                    SUM(
                        CASE
                            WHEN r.status = '已報名'
                            THEN r.people
                            ELSE 0
                        END
                    ), 0
                ) AS registered_people
            FROM events e
            LEFT JOIN registrations r
                ON e.id = r.event_id
            GROUP BY
                e.id, e.name, e.event_date, e.capacity
            ORDER BY e.id
        """)

        rows = cursor.fetchall()

        if not rows:
            return "目前沒有活動資料。"

        result = "===== 各活動報名統計 =====\n"

        for name, date, capacity, registered in rows:
            remaining = capacity - registered

            result += (
                f"\n活動：{name}\n"
                f"日期：{date}\n"
                f"總名額：{capacity} 人\n"
                f"已報名：{registered} 人\n"
                f"剩餘名額：{remaining} 人\n"
            )

        return result

    except sqlite3.Error as e:
        print(f"統計查詢錯誤：{e}")
        return "活動統計查詢失敗。"

    finally:
        conn.close()


# 2. 建立活動報名主函式
def activity_system(student_name, message):

    # 檢查輸入內容
    if not isinstance(message, str) or not message.strip():
        return "請輸入活動需求！"

    message = message.strip()

    # AI 辨識動作、活動、人數
    action, event, people, raw_reply = ai_parse(message)

    print(f"AI 解讀：{action}, {event}, {people}")

    # 無法辨識動作
    if action == "未知":
        return (
            "無法辨識您的需求。\n"
            "目前活動：迎新晚會、程式工作坊、籃球友誼賽。"
        )

    # A3：查詢全部活動統計，不需要姓名
    if action == "查詢" and event == "全部活動":
        return event_statistics()

    # 檢查活動是否合法
    if event not in EVENT_NAMES:
        return (
            "找不到指定活動。\n"
            "目前活動：迎新晚會、程式工作坊、籃球友誼賽。"
        )

    # 報名和取消需要學生姓名
    if action in ["報名", "取消"]:
        if not isinstance(student_name, str) or not student_name.strip():
            return "請先輸入學生姓名！"

        student_name = student_name.strip()

    # A1：報名人數只能是1～3人
    if action == "報名":
        if not isinstance(people, int) or not 1 <= people <= 3:
            return "報名失敗：每次只允許報名 1～3 人！"

    conn = sqlite3.connect(DB_NAME, timeout=10)

    try:
        conn.execute("PRAGMA foreign_keys = ON")
        cursor = conn.cursor()

        # 報名及取消使用交易，避免多人同時更新名額
        if action in ["報名", "取消"]:
            conn.execute("BEGIN IMMEDIATE")

        # 查詢活動是否存在
        cursor.execute(
            """
            SELECT id, name, capacity
            FROM events
            WHERE name = ?
            """,
            (event,)
        )

        row = cursor.fetchone()

        if row is None:
            return f"找不到活動：{event}"

        event_id, event_name, capacity = row

        # 3. 計算有效報名總人數（A1）
        cursor.execute(
            """
            SELECT COALESCE(SUM(people), 0)
            FROM registrations
            WHERE event_id = ?
              AND status = '已報名'
            """,
            (event_id,)
        )

        count_row = cursor.fetchone()
        registered = count_row[0] if count_row else 0

        remaining = capacity - registered

        # 4. 查詢剩餘名額
        if action == "查詢":
            return (
                f"活動：{event_name}\n"
                f"總名額：{capacity} 人\n"
                f"已報名：{registered} 人\n"
                f"剩餘名額：{remaining} 人"
            )

        # 5. 檢查學生是否有報名紀錄
        cursor.execute(
            """
            SELECT id, status
            FROM registrations
            WHERE event_id = ?
              AND student_name = ?
            """,
            (event_id, student_name)
        )

        record = cursor.fetchone()

        # 6. 取消報名功能
        if action == "取消":

            if record is None:
                return f"{student_name} 沒有報名過「{event_name}」。"

            record_id, record_status = record

            if record_status != "已報名":
                return f"{student_name} 的報名已取消，無法重複取消。"

            cursor.execute(
                """
                UPDATE registrations
                SET status = '已取消'
                WHERE id = ?
                  AND status = '已報名'
                """,
                (record_id,)
            )

            if cursor.rowcount != 1:
                conn.rollback()
                return "取消失敗，請重新操作。"

            conn.commit()

            return (
                f"取消成功！\n"
                f"學生：{student_name}\n"
                f"活動：{event_name}\n"
                f"已釋出原本的報名名額。"
            )

        # 7. 報名功能
        if action == "報名":

            # 不可重複報名
            if record is not None and record[1] == "已報名":
                return f"{student_name} 已報名過「{event_name}」，不可重複報名！"

            # A2：每位學生最多同時報名2個活動
            cursor.execute(
                """
                SELECT COUNT(DISTINCT event_id)
                FROM registrations
                WHERE student_name = ?
                  AND status = '已報名'
                """,
                (student_name,)
            )

            count_row = cursor.fetchone()
            active_events = count_row[0] if count_row else 0

            if active_events >= 2:
                return (
                    f"報名失敗：{student_name} 已報名 "
                    "2 個活動，不能再報名第三個活動！"
                )

            # A1：檢查剩餘名額
            if people > remaining:
                return (
                    f"報名失敗：「{event_name}」名額不足！\n"
                    f"目前剩餘：{remaining} 人\n"
                    f"本次需要：{people} 人"
                )

            # 如果之前取消過，重新啟用原紀錄
            if record is not None and record[1] == "已取消":

                cursor.execute(
                    """
                    UPDATE registrations
                    SET status = '已報名',
                        people = ?,
                        created_at = CURRENT_TIMESTAMP
                    WHERE id = ?
                      AND status = '已取消'
                    """,
                    (people, record[0])
                )

            else:
                # 第一次報名：新增紀錄
                cursor.execute(
                    """
                    INSERT INTO registrations
                    (event_id, student_name, people, status)
                    VALUES (?, ?, ?, ?)
                    """,
                    (event_id, student_name, people, "已報名")
                )

            # 確認資料真的更新成功
            if cursor.rowcount != 1:
                conn.rollback()
                return "報名失敗，資料未成功更新。"

            conn.commit()

            new_remaining = remaining - people

            return (
                f"報名成功！\n"
                f"學生：{student_name}\n"
                f"活動：{event_name}\n"
                f"報名人數：{people} 人\n"
                f"剩餘名額：{new_remaining} 人"
            )

        return "無法處理此動作。"

    except sqlite3.Error as e:
        conn.rollback()
        print(f"資料庫錯誤：{e}")
        return "系統發生資料庫錯誤，請稍後再試。"

    finally:
        # 確保交易結束並關閉連線
        if conn.in_transaction:
            conn.rollback()
        conn.close()
````

````python
# ============================================================
# 第 7 格：主流程與流程測試
# 提醒：流程測試 8～10 步；最後呼叫 show_tables()
# ============================================================
# （把 AI 產生、你看過並理解的程式貼在這裡）

# ============================================================
# 第 7 格：系學會活動報名系統－流程測試（9 步）
# ============================================================

# 1. 建立測試案例
# 格式：(學生姓名, 使用者輸入, 預期關鍵字)

flow_tests = [
    # 測試1：正常報名（A1 攜伴2人）
    (
        "小明",
        "我和 1 個朋友想報名迎新晚會。",
        "報名成功"
    ),

    # 測試2：重複報名
    (
        "小明",
        "我要再報名迎新晚會。",
        "不可重複報名"
    ),

    # 測試3：查詢剩餘名額
    (
        "小華",
        "迎新晚會還剩多少名額？",
        "剩餘名額：4 人"
    ),

    # 測試4：報名第二個活動
    (
        "小明",
        "我想報名程式工作坊。",
        "報名成功"
    ),

    # 測試5：超過兩活動限制（A2）
    (
        "小明",
        "我要報名籃球友誼賽。",
        "不能再報名第三個活動"
    ),

    # 測試6：攜伴3人，額滿（A1）
    (
        "小華",
        "我想和兩個朋友報名籃球友誼賽。",
        "報名成功"
    ),

    # 測試7：名額不足
    (
        "小美",
        "我想報名籃球友誼賽。",
        "名額不足"
    ),

    # 測試8：取消報名
    (
        "小華",
        "我要取消籃球友誼賽的報名。",
        "取消成功"
    ),

    # 測試9：全部活動統計（A3）
    (
        "幹部",
        "查詢各活動報名狀況。",
        "各活動報名統計"
    )
]


# 2. 初始化測試統計
passed = 0
total = len(flow_tests)
flow_results = []

print("=" * 55)
print("       系學會活動報名系統：9 步流程測試")
print("=" * 55)


# 3. 依序執行所有測試
for i, (student_name, message, expected) in enumerate(
    flow_tests, start=1
):

    print(f"\n===== 第 {i}/{total} 步 =====")
    print(f"學生姓名：{student_name}")
    print(f"使用者輸入：{message}")

    try:
        # 呼叫第6格核心函式
        result = activity_system(student_name, message)

        print(f"\n系統回應：\n{result}")

        # 4. 檢查是否符合預期關鍵字
        is_pass = (
            isinstance(result, str)
            and expected in result
        )

        if is_pass:
            passed += 1
            status = "PASS"
        else:
            status = "FAIL"

    except Exception as e:
        result = f"執行錯誤：{type(e).__name__}: {e}"
        status = "FAIL"
        print(result)

    print(f"\n預期關鍵字：{expected}")
    print(f"測試結果：{status}")
    print("-" * 55)

    # 5. 儲存測試結果
    flow_results.append({
        "步驟": i,
        "學生姓名": student_name,
        "使用者輸入": message,
        "預期關鍵字": expected,
        "測試結果": status
    })


# 6. 顯示流程測試總結
print("\n" + "=" * 55)
print("                 流程測試總結")
print("=" * 55)

print(f"測試總數：{total}")
print(f"通過數量：{passed}")
print(f"失敗數量：{total - passed}")
print(f"通過率：{passed / total * 100:.1f}%")


# 7. 顯示測試結果表格
flow_df = pd.DataFrame(flow_results)

print("\n===== 流程測試結果總表 =====")
display(flow_df)


# 8. 顯示最後的活動統計（A3）
print("\n===== 最終活動統計 =====")
print(event_statistics())


# 9. 顯示資料庫兩張資料表
print("\n===== 最終資料庫內容 =====")
show_tables()
````
